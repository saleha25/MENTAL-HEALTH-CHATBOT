```python
import os
os.chdir(r"C:\Users\tellm\Desktop\Mental Health Screening & Well-Being Assessment .csv")

from datasets import load_dataset
import pandas as pd

os.makedirs("data/external", exist_ok=True)

# 1. EmpatheticDialogues
print("⏳ 1/5 Downloading EmpatheticDialogues...")
try:
    ed = load_dataset("facebook/empathetic_dialogues")
    ed["train"].to_pandas().to_csv("data/external/empathetic_dialogues.csv", index=False)
    print(f"✅ Saved — {len(ed['train'])} rows")
except Exception as e:
    print(f"❌ Failed: {e}")

# 2. Reddit Mental Health
print("\n⏳ 2/5 Downloading Reddit Mental Health...")
try:
    rmh = load_dataset("solomonk/reddit_mental_health_posts")
    rmh["train"].to_pandas().to_csv("data/external/reddit_mental_health.csv", index=False)
    print(f"✅ Saved — {len(rmh['train'])} rows")
except Exception as e:
    print(f"❌ Failed: {e}")

# 3. CounselChat
print("\n⏳ 3/5 Downloading CounselChat...")
try:
    cc = load_dataset("nbertagnolli/counsel-chat")
    cc["train"].to_pandas().to_csv("data/external/counsel_chat.csv", index=False)
    print(f"✅ Saved — {len(cc['train'])} rows")
except Exception as e:
    print(f"❌ Failed: {e}")

# 4. MentalChat16K
print("\n⏳ 4/5 Downloading MentalChat16K...")
try:
    mc = load_dataset("ShenLab/MentalChat16K")
    mc["train"].to_pandas().to_csv("data/external/mental_chat_16k.csv", index=False)
    print(f"✅ Saved — {len(mc['train'])} rows")
except Exception as e:
    print(f"❌ Failed: {e}")

# 5. Mental Health Counseling
print("\n⏳ 5/5 Downloading Mental Health Counseling...")
try:
    mhc = load_dataset("Amod/mental_health_counseling_conversations")
    mhc["train"].to_pandas().to_csv("data/external/mental_health_counseling.csv", index=False)
    print(f"✅ Saved — {len(mhc['train'])} rows")
except Exception as e:
    print(f"❌ Failed: {e}")

# Summary
print("\n📁 Files saved:")
for f in os.listdir("data/external"):
    size = os.path.getsize(f"data/external/{f}") // 1024
    print(f"   ✅ {f}  —  {size} KB")

print("\n🎉 Done!")
```
